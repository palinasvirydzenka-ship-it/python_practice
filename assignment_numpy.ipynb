**1.** Given a random array, change the sign of elements whose values ​​are between 3 and 8

```python
ran_arr = np.random.randint(0, 10, size=12)
print("Random array:", ran_arr)
```

```python
mask = (ran_arr > 3) & (ran_arr < 8)

ran_arr[mask] *= -1
print("Changed array:", ran_arr)
```

**2.** Replace the maximum element of a random array with 0

```python
ran_arr[ran_arr.argmax()] = 0
print("Modified array:", ran_arr)
```

**3.** Construct the direct product of arrays (all combinations with each element). A two-dimensional array is provided as input.

```python
input_array = np.array([[1, 4], [8, 5], [3, 2]])

combinations = []

for x in input_array[0]:  
    for y in input_array[1]:  
        for z in input_array[2]:  
            
            combinations.append([x, y, z])

results = np.array(combinations)

print("Input array:\n", input_array)
print("\nThe results of the combinations\n", results)
```

**4.** Given two arrays, A (8x3) and B (2x2), find rows in A that contain elements from each row in B, regardless of the order of the elements in B.

```python
A = np.random.randint(1, 10, size=(8, 3))
B = np.random.randint(1, 10, size=(2, 2))

print("Array A (8x3):\n", A)
print("\nArray B (2x2):\n", B)
```

```python
mask_B1 = np.isin(A, B[0]).any(axis=1)
mask_B2 = np.isin(A, B[1]).any(axis=1) 

final_mask = mask_B1 & mask_B2
result = A[final_mask]
print("\nThe final result\n", result)
```

**5.** Given a 10x3 matrix, find rows of unequal values ​​(for example, the row [2,2,3] remains, the row [3,3,3] is deleted)

```python
matrix = np.array([[1,1,1], [3, 3, 3], [5, 1, 5], [7, 7, 7], [1, 2, 3],
[7,10,5], [4, 4, 4], [8, 8, 2], [0, 0, 0], [6, 5, 6]
])

all_equal_mask = np.all(matrix == matrix[:, [0]], axis=1)

unequal_mask = ~all_equal_mask
final_result = matrix[unequal_mask]

print("The final results", final_result)
```

* __Task 1__: Calculate the product of nonzero elements on the diagonal of a rectangular matrix.  
 For example, for X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) answer is 3.

```python
X_list = [[0, 0, 1], [2, 3, 2], [3, 6, 3], [4, 1, 4]]
def diag_nonzero_python(X):
    import math

    diag_1 = [X[i][i] for i in range(min(len(X), len(X[0])))]
    return math.prod([x for x in diag_1 if x != 0])

print("Python:", diag_nonzero_python(X_list)) 
```

```python
X_np = [[2, 0, 1], [2, 1, 2], [3, 0, 3], [4, 0, 4]]
def prod_diag_nonzero_numpy(X):
    diag_2 = np.diag(X)
    return np.prod(diag_2[diag_2 != 0])

print("NumPy:", prod_diag_nonzero_numpy(X_np))  
```

* __Task 2__: Two vectors x and y are given. Check if they define the same multiset.  
  For example, for x = np.array([1, 2, 2, 4]), y = np.array([4, 2, 1, 2]) answer is True.

```python
x_list_1 = [1, 3, 2, 4]
y_list_1 = [4, 2, 1, 2]

def is_same_multiset_python(x, y):
    return sorted(x) == sorted(y)
print("Python:", is_same_multiset_python(x_list_1, y_list_1))
```

```python
x_np_1 = np.array(x_list_1)
y_np_1 = np.array(y_list_1)

def is_same_multiset_numpy(x, y):
    return np.array_equal(np.sort(x), np.sort(y))
print("NumPy:", is_same_multiset_numpy(x_np_1, y_np_1)) 
```

* __Task 3__: Find the maximum element in the vector x among the elements preceded by zero. 
 For example, for x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) answer is 5.

```python
x_list_2 = [6, 0, 1, 3, 2, 0, 5, 0, 8]

def max_after_zero_python(x):
    candidates = [x[i] for i in range(1, len(x)) if x[i - 1] == 0]
    return max(candidates) if candidates else None
print("Python:", max_after_zero_python(x_list_2))  # Output: 5
```

```python
x_np_2 = np.array(x_list_2)

def max_after_zero_numpy(x):
    zero_mask = x[:-1] == 0
    return np.max(x[1:][zero_mask]) if zero_mask.any() else None
print("NumPy:", max_after_zero_numpy(x_np_2))
```

* __Task 4__: Implement run-length encoding. For some vector x, it is necessary to return a tuple of two vectors of the same length. The first one contains numbers and the second one contains how many times they need to be repeated.  
 For example, for x = np.array([2, 2, 2, 3, 3, 3, 5]) answer is (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
from itertools import groupby

x_list_3 = [2, 5, 5, 5, 3, 3, 5]

def rle_python(x):
    keys, groups = zip(*[(k, sum(1 for _ in g)) for k, g in groupby(x)])
    return list(keys), list(groups)

val_py, cnt_py = rle_python(x_list_3)
print("Python:", val_py, cnt_py)
```

```python
x_np_3 = np.array([2, 5, 5, 5, 3, 3, 5])


def rle_numpy(x):
    end_indices = np.append(np.where(x[1:] != x[:-1])[0], len(x) - 1)
    return x[end_indices], np.diff(np.append(-1, end_indices))

val_np, cnt_np = rle_numpy(x_np_3)
print("NumPy:", val_np, cnt_np)
```

* __Task 5__: Two samples of objects are given - X and Y. Calculate the matrix of Euclidean distances between objects. Compare with function scipy.spatial.distance.cdist by operating speed.

```python
import math

def cdist_python(X, Y):
    return [
        [
            math.sqrt(sum((x - y) ** 2 for x, y in zip(row_x, row_y)))
            for row_y in Y
        ]
        for row_x in X
    ]
```

```python
def cdist_numpy(X, Y):

    return np.sqrt(np.sum((X[:, np.newaxis, :] - Y[np.newaxis, :, :]) ** 2, axis=2))
```

```python
import time
from scipy.spatial.distance import cdist


X_list_4 = np.random.rand(100, 50).tolist()
Y_list_4 = np.random.rand(100, 50).tolist()

X_np_4 = np.array(X_list_4)
Y_np_4 = np.array(Y_list_4)

# 1. Python
start = time.time()
res_py = cdist_python(X_list_4, Y_list_4)
time_py = time.time() - start

# 2. NumPy
start = time.time()
res_np = cdist_numpy(X_np_4, Y_np_4)
time_np = time.time() - start

# 3. SciPy
start = time.time()
res_scipy = cdist(X_np_4, Y_np_4)
time_scipy = time.time() - start

print(f"Python: {time_py:.4f} sec")
print(f"NumPy:  {time_np:.4f} sec")
print(f"SciPy:  {time_scipy:.4f} sec")

print("Results matched:", np.allclose(res_np, res_scipy))
```
